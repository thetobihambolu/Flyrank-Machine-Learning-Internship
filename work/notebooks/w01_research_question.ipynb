## 4. Careful words: what I can and can't claim

*Write what your work will be able to say (observed, directional, decision-support) — and what it never will (causal proof, 'predicting Google').*

## What This Work Can and Cannot Claim

**What it will say: observed**
This analysis reports what has already happened in the data — actual positions, actual clicks, actual trends over the observed window. When we say "38% of content sits in positions 8–20," that is a description of the current state.

**What it will say: directional**
Where we identify patterns — declining trend, stale content, near-miss positioning — we are pointing at *direction of movement*, not precise magnitude of future outcome. 
"This page is losing ground" is a directional claim. "This page will gain exactly 6 positions if refreshed" is not something this work can support.

**What it will say: decision-support**
The output is a prioritized queue with reasons attached — designed to help a human content team decide where to spend limited time next. It surfaces candidates and evidence; it does not make the final call. The team's judgment, resourcing, and execution quality remain the deciding factors in outcome.

---

**What it will never say: causal proof**
Correlation between a feature (e.g., content age, word count, freshness) and performance is not evidence that changing that feature *causes* a performance change. Pages with strong signals may simply come from stronger domains, better topics, or better original writing — confounds this analysis does not control for. 
We are not running a controlled experiment (e.g., an A/B test on refreshed vs. non-refreshed pages), so no causal claim ("refreshing X caused Y% traffic lift") 
is supportable from this data alone.

**What it will never say: "predicting Google"**
This work does not model or claim to predict Google's ranking algorithm, its weighting of signals, or future algorithm changes. Google's ranking system is a black box with hundreds of undisclosed, frequently-updated factors. Any pattern found here is a *correlation observed in this client's historical data*, not a discovered ranking rule, and it may not generalize to other sites, other verticals, or even the same site after the next algorithm update.
